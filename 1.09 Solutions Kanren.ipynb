# Kanren Exercises

1. Write kanren functions to implement the material implication ($\implies$) and bidirectional implication ($\iff$) operators (hint: use the truth tables to help you with this).

```python
from kanren import lany, eq, var, run, lall
# Material implication
def implies(A,B):
    # A => B can be written as not(A) or B
    return lany(eq(A,False),eq(B,True))

def iff(A,B):
    # A <=> B can be written as A xnor B which can be written as (A^B)v(not(A)^not(B))
    return lany(lall(eq(A,True),eq(B,True)), lall(eq(A,False),eq(B,False)))
```


2. Write Kanren programs to verify the following arguments

a)  $P\implies Q,\ P\ : Q$

```python
from kanren import vars
P,Q = vars(2)
goals = lall(implies(P,Q), eq(P, True))
run(0, Q, goals)
```

b)  $P\implies Q,\ \lnot Q\ : \lnot Q$

```python
goals = lall(implies(P,Q), eq(Q, False))
run(0, P, goals)
```

c)  $P\implies Q,\ Q\implies R,\ P\ : R$

```python
R = var()
goals = lall(implies(P,Q), implies(Q,R), eq(P, True))
run(0, R, goals)
```

d)  $P\implies Q,\ Q\implies R,\ \lnot R\ : \lnot P$

```python
goals = lall(implies(P,Q), implies(Q,R), eq(R, False))
run(0, P, goals)
```

e) $P\land Q,\ P\implies R\ :\ R$

```python
def land(A,B):
    return lall(eq(A,True), eq(B,True))

goals = lall(land(P,Q), implies(P,R))
run(0, R, goals)
```

f) $P\lor Q,\ P\implies R,\ \lnot Q\ :\ R$

```python
def lor(A,B):
    return lany(eq(A,True), eq(B,True))

goals = lall(lor(P,Q), implies(P,R), eq(Q, False))
run(0, R, goals)
```

g) $P\implies Q\land R,\ R\implies S,\ P,\ Q\ :\ S$

```python
P,Q,R,S = vars(4)
goals = lall(
    lany(eq(P,False),lall(eq(Q,True),eq(R,True))),
    lany(eq(R,False),eq(S,True)),
    eq(P, True),
    eq(Q, True)
)

run(0, S, goals)
```

h) $\lnot P\implies \lnot Q,\ Q,\ P\implies S\land Q\ :\ S $

```python
goals = lall(
    lany(eq(P,True),eq(Q,False)),
    eq(Q, True),
    lany(eq(P,False),lall(eq(S,True),eq(Q,True)))
)
run(0, S, goals)
```

i) $(P\land Q)\implies R,\ \lnot R,\ P\ :\ \lnot Q$

```python
goals = lall(
    lany(lany(eq(P,False),eq(Q,False)),eq(R,True)),
    eq(R, False),
    eq(P, True)       
)

run(0, Q, goals)
```

j) $P\land Q,\ Q\implies\lnot R\ :\ \lnot(R\land\lnot S)$

```python
goals=(
    lall(eq(P,True),eq(Q,True)),
    lany(eq(Q,False),eq(R,False)),
    lany(eq(R,False),eq(S,True))
)
run(0,[P,Q,R,S],goals)
```

k) $\lnot P,\ \lnot Q\ :\ P\iff Q$

```python
goals=(
    eq(P,False),
    eq(Q,False),
    lany(lall(eq(P,True),eq(Q,True)),lall(eq(P,False),eq(Q,False)))
)
run(0,[P,Q,R,S],goals)
```

l) $P\implies \lnot Q \land\lnot R,\ P\ :\ \lnot Q$

```python
goals = (
    lany(eq(P,False),lall(eq(Q,False),eq(R,False))),
    eq(P,True),
    eq(Q,False)
)
run(0,[P,Q,R],goals)
```

m) Write a kanren program to verify the argument $P\implies Q,\ \lnot Q\ :\ P\implies R$. You may find the following equivalence helpful: $A\implies B \equiv \lnot A \lor B$.

```python
goals=(
    lany(eq(P,False),eq(Q,True)),
    eq(Q,False),
    lany(eq(P,False),eq(R,True))
)
run(0, [P,Q,R], goals)
```

2. Write kanren programs to verify that the following argument are invalid:

a) $\lnot(P\land Q)\implies S,\ \lnot S,\ P\implies\lnot Q\ :\ \lnot Q$

```python
goals = lall(
    lany(lall(eq(P,True), eq(Q,True)), eq(S,True)),
    eq(S, False),
    lany(eq(P, False), eq(Q, False))
)
run(0, Q, goals)
```

b) Write a kanren program to verify that the following argument is invalid: $P\iff Q,\ P,\ Q\implies\lnot R,\ R\ :\ Q$

```python
goals=(
    lany(lall(eq(P,True),eq(Q,True)), lall(eq(P,False),eq(Q,False))),
    eq(P,True),
    lany(eq(Q,False),eq(R,False)),
    eq(R,True)
)
run(0, Q, goals)
```

3. The following problems are more challenging examples. It is unlikely that your function for `implies` or similar functions for `and` and `or` will be useful here and you will have to think carefully about how to formulate these.

**You are not expected to be able to solve these, but they are included here for the most enthusiastic students.**

a) $P\implies Q\land R,\ P\ :\ R$

[Hint: kanren may not be able to "reason through" your `implies` function here, try representing the logic explicitly]

```python
goals = lall(
    lany(eq(P,False), lall(eq(Q,True), eq(R,True))),
    eq(P, True)
)
run(0, R, goals)
```

b) $P\implies Q\lor R,\ P\ , \lnot Q:\ R$

```python
goals = lall(
    lany(eq(P,False), lany(eq(Q,True), eq(R,True))),
    eq(P, True),
    eq(Q,False)
)
run(0, R, goals)
```

c) $P\lor Q \implies R,\ \lnot R\ : \lnot P$

[This is a especially tricky problem. Hint: think about when $P\lor Q$ is False and use this to simplify the goal]

```python
goals = lall(
    lany(lall(eq(P,False),eq(Q,False)), eq(R, True)),
    eq(R, False)
)
run(0, P, goals)
```
