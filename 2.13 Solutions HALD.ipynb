1. Write a SPARQL Query to find the node with the highest number of outgoing relations.

```python
x = list(default_world.sparql(
    """
    PREFIX ONTO: <http://www.dummy.info/new.owl#>
    
    SELECT ?node (COUNT(*) AS ?incoming)
    WHERE {
    ?node ?p ?o
    }
    GROUP BY ?node
    ORDER BY DESC(?incoming)
    """))


for i in x:
    print(i)
```

2. Write a SPARQL Query to find the node with the highest number of outgoing relations. You will need to use at least on `FILTER` to remove the attributes from the results.

```python
x = list(default_world.sparql(
    """
    PREFIX ONTO: <http://www.dummy.info/new.owl#>
    
    SELECT ?node (COUNT(*) AS ?incoming)
    WHERE {
    ?s ?p ?node
    FILTER (!isLiteral(?node))
    FILTER (?node != owl:NamedIndividual)
    }
    GROUP BY ?node
    ORDER BY DESC(?incoming)
    """))


for i in x:
    print(i)
```

3. Write a SPARQL query to find all paths of length 2 starting at node "CD4" and ending at node "Death"

```python


x = list(default_world.sparql(
    """
    PREFIX ONTO: <http://www.dummy.info/new.owl#>
    
    SELECT ?node1 ?r1 ?mid ?r2 ?node2
    WHERE {
        ?node1 ONTO:entity "CD4"
        ?node2 ONTO:entity "Death"
        ?node1 ?r1 ?mid
        ?mid ?r2 ?node2
    }
    """))


for i in x:
    print(i)
```

4. Find the set of distinct nodes that connect CD4 and Death. Report their `entity` sorted alphabetically.

```python

x = list(default_world.sparql(
    """
    PREFIX ONTO: <http://www.dummy.info/new.owl#>
    
    SELECT DISTINCT ?midname
    WHERE {
        ?node1 ONTO:entity "CD4"
        ?node2 ONTO:entity "Death"
        ?node1 ?r1 ?mid
        ?mid ?r2 ?node2
        ?mid ONTO:entity ?midname
    }
    ORDER BY ?midname
    """))


for i in x:
    print(i)
```

5. Write a SPARQL Query to find all `Protein`s that have a direct relation with a `Lipid`

```python
# this code can take a couple of minutes to run - be patient
x = list(default_world.sparql(
    """
    PREFIX ONTO: <http://www.dummy.info/new.owl#>

    SELECT ?protein ?relation ?lipid ?direction
    WHERE {
      {
        ?protein a ONTO:Protein
        ?lipid   a ONTO:Lipid
        ?protein ?relation ?lipid
        BIND("protein -> lipid" AS ?direction)
      }
      UNION
      {
        ?protein a ONTO:Protein
        ?lipid   a ONTO:Lipid
        ?lipid ?relation ?protein
        BIND("lipid -> protein" AS ?direction)
      }
    }
    GROUP BY ?direction ?protein ?relation ?lipid
    """))
    
    
for i in x:
  print(i)
```
