```python
import ollama

# Send a prompt to the local model
response = ollama.chat(model='llama3.2:3b', messages=[
  {
    'role': 'user',
    'content': 'What are the main benefits of running LLMs locally?',
  },
])

# Print the model's response
print(response['message']['content'])
```
