# A Note on Namespaces and Prefixes

We have used namespace, usually referred to using a shorthand prefix, liberally but rather casually. It is helpful to clarify what they refer to and how we use them.

OWL is a specialisation of the RDF data model. In RDF, every resource (class, property or individual) is identified by a full "IRI", such as http://www.w3.org/2002/07/owl#NamedIndividual. Writing these in full makes queries long and hard to read. Namespaces solve this with prefixes. A PREFIX declaration binds a short label to a base IRI, and the query can then use compact names like `owl:DataProperty` . The parser expands each compact name by joining the prefix’s IRI with the local part, so `owl:DataProperty` and `<http://www.w3.org/2002/07/owl#DataProperty>` are identical.

We will typically use three namespaces in our examples.

* http://www.w3.org/1999/02/22-rdf-syntax-ns# is the IRI for the standard RDF data model. Resources such as `type` are defined here. The prefix `rdf` is pre-assigned to this in `owlready2` and so we are able to refer to `rdf:type` without explicitly defining the prefix.

* http://www.w3.org/2002/07/owl# is the IRI for the OWL extension to th RDF model. It is here that resources such as `Class`, and `ObjectProperty` are defined. 

* http://www.dummy.info/new.owl# is the IRI for the model that we have created using this IRI as a base. It is here that resources such as `Module` and `Passed` are defined.

We can use all of these namespaces in a single query, and can define separate prefixes for them. In the example below, we use all three, and redefine the `rdf` prefix for clarity (we could omit the definition as it is already defined in the software package).

We could also use other namespaces which would allow us to refer to other resources. We might do this if we were working with a pre-existing ontology that we wanted to use to structure our graph. 

Note that the keyword `a` is a shorthand for `rdf:type`.

```python
from owlready2 import *
onto = get_ontology("teaching.owl").load()

x = list(default_world.sparql(
    """
    PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
    PREFIX owl: <http://www.w3.org/2002/07/owl#>
    PREFIX onto: <http://www.dummy.info/new.owl#>
    
    SELECT ?s ?r ?o 
    WHERE{
        ?s ?r ?o
        ?r a owl:ObjectProperty
        ?o rdf:type onto:Module
        ?s a onto:Student
    }
    """
))


for i in x:
    print(i)
```
